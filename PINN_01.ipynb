**Tensor Maths Operation**

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])
tensor_a
```

```python
# do some addition
tot = tensor_a + tensor_b
tot
```

```python
# Using tensor function for additon
pop = torch.add(tensor_a,tensor_b)
pop
```

```python
# subtraction
tensor_b - tensor_a
```

```python
# Using the fund=ction for subtraction
torch.sub(tensor_b,tensor_a)
```

```python
#using the multiplication  is also natural
tensor_a * tensor_b
```

```python
Mup = torch.mul(tensor_a,tensor_b)
Mup
```

```python
#Division is a little trickier
tensor_b/tensor_a
```

```python
#using the div function
dip = torch.div(tensor_b,tensor_a)
dip
```

```python
# Reminder or Modulus
tensor_b % tensor_a
```

```python
# Reminder function
rep = torch.remainder(tensor_b,tensor_a)
rep
```

```python
# Exponetial / Power
Por = torch.pow(tensor_a,tensor_b)
Por
```

```python
# Another way to add
tensor_a.add_(tensor_b)
```
